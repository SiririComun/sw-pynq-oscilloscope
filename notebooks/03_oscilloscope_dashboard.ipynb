# Notebook 3: Real-Time Multi-Tab Oscilloscope & Spectrum Analyzer Dashboard

This notebook deploys the complete interactive **Multi-Domain Instrument** using `OscilloscopeOverlay`.

### Features:
- 📈 **Oscilloscope Tab:** Real-time 1 MSPS analog waveform with sub-microsecond hardware edge triggering and live threshold line.
- 📊 **Spectrum Analyzer Tab:** Real-time 2048-point FPGA FFT ($0 - 500\,\text{kHz}$) with auto peak detector ($f_0$) and configurable dBV/dBFS scales.
- 🔀 **Dual View Tab:** Simultaneous split view showing Time and Frequency domains concurrently.
- ⚡ **AD3 Signal Generator:** Non-blocking background waveform generation (Sine, Triangle, Square).

## 2. Launch Complete Multi-Tab Dashboard
Instantiate `OscilloscopeOverlay` and call `.dashboard()` to start the real-time application.

```python
from pynq_oscilloscope import OscilloscopeOverlay

# 1. Initialize Overlay (downloads v1.2.0-rc1 or loads local .bit)
ol = OscilloscopeOverlay()

# 2. Render and launch interactive Dashboard
app = ol.dashboard()
```

## 3. Hardware Shutdown & Cleanup
Click **Stop** on the dashboard or execute this cell to disconnect the AD3 and release memory.

```python
app.stop()
ol.close()
print("Dashboard stopped and memory released.")
```
