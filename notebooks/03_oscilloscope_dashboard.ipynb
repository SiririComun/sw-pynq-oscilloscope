# Notebook 3: Real-Time Dual-Channel Oscilloscope & Spectrum Analyzer Dashboard

This notebook deploys the complete interactive **Multi-Domain Instrument** supporting simultaneous dual-channel acquisition on the PYNQ-Z2 (`v1.3.0-rc1`).

### 🌟 Multi-Channel Instrument Features:
- 📈 **Dual Oscilloscope Tab:** Real-time 1 MSPS synchronous traces for **CH1 (A0, Cyan)** and **CH2 (A1, Magenta)** with hardware edge triggering and live threshold line.
- 📊 **Spectrum Analyzer Tab:** Real-time 2048-point FPGA FFT ($0 - 500\,\text{kHz}$) with auto peak detector ($f_0$) and configurable dBV/dBFS/Linear scales.
- 🔀 **Dual View Tab:** Stacked multi-domain view displaying both Time-Domain channels (top) and Frequency-Domain spectrum (bottom) concurrently at ~30 FPS.
- ⚡ **Dual AD3 Wavegen Controls:** Real-time independent waveform, frequency, and amplitude controls for **W1 (A0)** and **W2 (A1)**.

## 1. System & USB Permission Verification

```python
from pynq_oscilloscope import check_usb_permissions, OscilloscopeOverlay

# Grant permissions for /dev/bus/usb to communicate with AD3
check_usb_permissions()
```

## 2. Launch Complete Multi-Tab Dual Instrument
Instantiate `OscilloscopeOverlay` and call `.dashboard()` to start the real-time application.

```python
# 1. Initialize Dual-Channel Overlay (loads v1.3.0-rc1 bitstream)
ol = OscilloscopeOverlay()
```

```python
# 2. Render and launch interactive Dashboard
app = ol.dashboard()
```

## 3. Hardware Shutdown & Cleanup
Click **Stop** on the dashboard or execute this cell to disconnect the AD3 and release hardware CMA memory.

```python
app.stop()
ol.close()
print("🔒 Dashboard stopped and memory released.")
```
