# Notebook 3: Real-Time Interactive Oscilloscope Dashboard

This notebook deploys the main interactive closed-loop **Oscilloscope Dashboard** using `pynq_oscilloscope`.

### Features:
- **Real-time 1 MSPS DMA Acquisition** (XADC via AXI DMA)
- **Non-blocking Waveform Generation** (Analog Discovery 3 via `pydwf` background thread)
- **Interactive Dark-Mode Plotly UI** with `uirevision` viewport locks
- **Zero-Latency Client-Side Controls** using `widgets.jslink()`
- **Software Rising-Edge Trigger** (Stationary Waveform Mode)
- **Adaptive Axis Auto-Ranging** & Live $V_{pp}$ Readout

## 1. System & USB Permission Verification
Ensure USB permissions are granted so Jupyter can communicate with the Analog Discovery 3.

```python
from pynq_oscilloscope import check_usb_permissions

# Set permissions for /dev/bus/usb
check_usb_permissions()
```

## 2. Automatic Hardware Overlay Loading
Fetch and load the pinned hardware overlay (`v1.0.2`) from GitHub Releases.

```python
from pynq_oscilloscope import HardwareLoader

# Auto-detect board model and load target overlay
overlay = HardwareLoader.load_overlay()
print("Hardware overlay loaded successfully!")
```

## 3. Deploy Interactive Oscilloscope Dashboard
Instantiate the high-level dashboard and display the UI controls and live Plotly canvas.

```python
from pynq_oscilloscope import OscilloscopeDashboard

# Instantiate the Dashboard
app = OscilloscopeDashboard(overlay=overlay)

# Display UI Layout & Plotly Canvas
app.display()
```

## 4. Hardware Shutdown & Cleanup
Run this cell or click **Stop** on the dashboard to safely disconnect the Analog Discovery 3 and release the DMA memory buffer.

```python
# Halt background update loop and release hardware handles
app.stop()
print("Oscilloscope dashboard stopped cleanly.")
```
