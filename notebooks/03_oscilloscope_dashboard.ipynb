# Notebook 3: Academic Dual-Channel Oscilloscope & Spectrum Analyzer Dashboard

This notebook deploys the interactive **Laboratory Instrument** with Analog Discovery 3 (AD3) integration on the PYNQ-Z2 (`v1.5.0`).

### 🌟 Academic Instrument Features:
- 📈 **500 kSPS Dual Scope:** Real-time synchronous traces for **CH1 (A0, Cyan)** and **CH2 (A1, Magenta)** with hardware edge triggering and live threshold line.
- ⚡ **Wideband AD3 Wavegen Controls ($10\text{ Hz} - 1\text{ MHz}$):** Real-time independent frequency, amplitude, and waveform controls for **W1 (A0)** and **W2 (A1)**.
- 🔬 **Nyquist Folding & Aliasing Exploration:** Freely dial frequencies beyond the $250\,\text{kHz}$ Nyquist limit ($300\,\text{kHz}, 450\,\text{kHz}, 800\,\text{kHz}$) to observe real-time spectral folding.
- 📊 **Dual Spectrum Analyzer Tab:** Real-time 2048-point spectrum ($0 - 250\,\text{kHz}$) with auto peak detector ($f_0$) and configurable dBV/dBFS/Linear scales.
- 🔀 **4-Tab Synchronized Display:** Dual Scope, Dual FFT, and isolated Channel 1/2 Views.

## 1. System & USB Permission Verification

```python
from pynq_oscilloscope import check_usb_permissions, OscilloscopeOverlay

# Grant permissions for /dev/bus/usb to communicate with AD3
check_usb_permissions()
```

## 2. Launch Complete Multi-Tab Dual Instrument
Instantiate `OscilloscopeOverlay` and call `.dashboard()` to start the real-time application.

```python
# 1. Initialize Dual-Channel Overlay (loads v1.3.0-rc1 bitstream)
ol = OscilloscopeOverlay()
```

```python
# 2. Render and launch interactive Dashboard
app = ol.dashboard()
```

## 3. Hardware Shutdown & Cleanup
Click **Stop** on the dashboard or execute this cell to disconnect the AD3 and release hardware CMA memory.
