## 2. Hardware Connection Verification
Test physical connectivity with the Analog Discovery 3.

```python
from pydwf import DwfLibrary
from pydwf.utilities import openDwfDevice

dwf = DwfLibrary()
try:
    with openDwfDevice(dwf) as device:
        print("=====================================================")
        print(" SUCCESS: Analog Discovery 3 is connected!")
        print("=====================================================")
except Exception as e:
    print(f"CONNECTION FAILED: {e}")
```

## 3. Non-Blocking Waveform Generation
Using `AD3SignalGenerator`, we launch signal generation in a background worker thread. This keeps the Jupyter kernel responsive for subsequent cells.

```python
import time
from pynq_oscilloscope import AD3SignalGenerator

# Instantiate high-level signal generator
wavegen = AD3SignalGenerator()

# Start generating a 10 kHz Sine wave (1.5V amplitude, 1.65V offset)
wavegen.start(shape="Sine", frequency=10000.0, amplitude=1.5, offset=1.65)

time.sleep(2.0)

# Dynamically update to a 25 kHz Square wave
print("Updating signal to 25 kHz Square wave...")
wavegen.update_parameters(shape="Square", frequency=25000.0, amplitude=1.2)

time.sleep(2.0)

# Clean hardware shutdown
wavegen.stop()
print("Wavegen stopped cleanly.")
```
