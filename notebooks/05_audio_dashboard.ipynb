# Notebook 5: Real-Time Dual-Channel Audio & Microphone Dashboard

This notebook deploys the dedicated **`AudioDashboard`** using `OscilloscopeOverlay` (`v1.4.0`).

Designed specifically for passive **MAX4466 electret microphones** (or any analog audio sensor on pins **A0** and **A1**), this dashboard runs completely independently without requiring an Analog Discovery 3 (AD3).

### 🌟 Audio Instrument Features:
- 🎙 **Passive Mode (Zero AD3 Dependency):** Starts instantly with no background worker threads or missing hardware warnings.
- ⏱ **Expanded $40.96\,\text{ms}$ Audio Timebase:** Displays multi-cycle acoustic waveforms for speech, musical instruments, and bass frequencies ($50\,\text{Hz} - 250\,\text{Hz}$).
- 🚨 **Real-Time VU Meters & Clipping Alerts:** Status bar indicators that flash red if either microphone saturates ($V < 0.10\,\text{V}$ or $V > 3.10\,\text{V}$).
- 🎯 **Sub-Bin Quadratic Peak Pitch Tracking:** Extracts the dominant acoustic fundamental ($f_0$) with $\pm 0.5\,\text{Hz}$ accuracy.
- 🪟 **Hann-Windowed Spectral Analysis:** Suppresses spectral leakage by $>32\,\text{dB}$ for clean harmonic overtone visualization.
- 🔀 **4-Tab Synchronized Display:** Dual Scope, Dual Audio FFT ($0 - 25\,\text{kHz}$), and dedicated Channel Views.
